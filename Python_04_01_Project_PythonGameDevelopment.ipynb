```python
# Hangman

# Author: Zaid Budi
# Language: Python
#------------------------------------------------------------------------------------------------------------------------------------------------------

# A text-based hangman game. The player picks a category and a difficulty, then guesses a hidden word one letter at a time before running out of lives.

#Features:
# - 3 word categories (football, countries, animals)
# - 3 difficulty levels (easy = 8 lives, medium = 6, hard = 4)
# - ASCII hangman drawing that builds up with each wrong guess
# - Input validation, so invalid choices and guesses never crash the game
# - Global counters tracking wins, losses, guesses, win rate and streaks
# - Game history and a final summary when the player stops
#------------------------------------------------------------------------------------------------------------------------------------------------------

# How to run: Kernel -> Restart & Run all, then follow the prompts
```

1. Setup: global counters, word bank and hangman drawings

```python
import random

#counters to track the player's statistics
games_played = 0
wins = 0
losses = 0
correct_guesses = 0
wrong_guesses = 0
current_streak = 0
best_streak = 0
game_history = []

```

```python

# Word bank: each category holds a list of words 
WORD_BANK = {
    "football": ["arsenal", "goalkeeper", "penalty", "striker", "offside", "stadium", "referee", "midfield", "corner", "tackle"],
    "countries": ["england", "brazil", "japan", "nigeria", "canada", "germany", "portugal", "egypt", "mexico", "australia"],
    "animals": ["elephant", "giraffe", "penguin", "kangaroo", "dolphin", "tiger", "zebra", "squirrel", "octopus", "rabbit"]
}

# Number of lives the player gets for each difficulty level
DIFFICULTY_LIVES = {"easy": 8, "medium": 6, "hard": 4}
```

```python
HANGMAN_STAGES = [
    """
  +---+
  |   |
      |
      |
      |
      |
=========""",
    """
  +---+
  |   |
  O   |
      |
      |
      |
=========""",
    """
  +---+
  |   |
  O   |
  |   |
      |
      |
=========""",
    """
  +---+
  |   |
  O   |
 /|   |
      |
      |
=========""",
    """
  +----+
  |    |
  O    |
 /|\\  |
       |
       |
=========""",
    """
  +----+
  |    |
  O    |
 /|\\  |
 /     |
       |
=========""",
    """
  +----+
  |    |
  O    |
 /|\\  |
 / \\  |
       |
========="""
]
```

2. Display functions

```python
def get_random_word(category):
    """Return a random word from the chosen category in the word bank."""
    words = WORD_BANK[category]
    return random.choice(words)
```

```python
def display_word(word, guessed_letters):
    """Return the word with guessed letters shown and blanks (_) for the rest."""
    display = []
    for letter in word:
        if letter in guessed_letters:
            display.append(letter)
        else:
            display.append("_")
    return " ".join(display)
```

```python
def draw_hangman(lives, max_lives):
    """Return the hangman drawing that matches how many lives have been lost.
    Scales the 7 drawing stages to the number of lives for the chosen difficulty."""
    lives_lost = max_lives - lives
    stage = round(lives_lost / max_lives * (len(HANGMAN_STAGES) - 1))
    return HANGMAN_STAGES[stage]
    # Works out how far through the drawing the player is (0 = empty, 6 = full body)
```

3. Game logic

```python
def play_round(word, lives):
    """Play one round of Hangman.
    Returns two values: True/False for win/loss, and the number of guesses used."""
    global correct_guesses, wrong_guesses  # lets this function update the global counters

    guessed_letters = set()  
    max_lives = lives  # remembers the starting lives so the drawing can be scaled

    while lives > 0: # Shows the current state of the game
        print(draw_hangman(lives, max_lives))  
        print("\nWord:", display_word(word, guessed_letters))
        print("Lives left:", lives)
        print("Letters guessed:", " ".join(sorted(guessed_letters)))

        guess = get_valid_guess(guessed_letters)
        guessed_letters.add(guess)

        if guess in word:
            print("Correct! '{}' is in the word.".format(guess))
            correct_guesses += 1
        else:
            print("Wrong! '{}' is not in the word.".format(guess))
            wrong_guesses += 1
            lives -= 1

        # If there's no blanks left, the whole word has been guessed
        if "_" not in display_word(word, guessed_letters):
            print("\nWell done! You guessed the word:", word.upper())
            return True, len(guessed_letters)

    # Only reached if the loop ends because lives hit 0
    print(draw_hangman(lives, max_lives))  
    print("\nUnlucky! You ran out of lives. The word was:", word.upper())
    return False, len(guessed_letters)
```

4. Input and validation functions

```python
def show_instructions():
    """Print the welcome message and the rules of the game."""
    print("=" * 40)
    print("        WELCOME TO HANGMAN")
    print("=" * 40)
    print("How to play:")
    print("1. Pick a category and a difficulty.")
    print("2. Guess the hidden word one letter at a time.")
    print("3. Every wrong guess costs you a life.")
    print("4. Guess the whole word before your lives run out to win!")
    print("Difficulty: Easy = 8 lives, Medium = 6 lives, Hard = 4 lives")
    print("=" * 40)
```

```python
def get_choice(prompt, options):
    """Ask the player to pick from a list of options and keep asking until
    they give a valid answer. Accepts either the option number or its name.
    Returns the chosen option as a lower-case string."""
    while True:
        print("\n" + prompt)
        for i in range(len(options)):
            print("  {}. {}".format(i + 1, options[i].capitalize()))

        choice = input("Enter a number or name: ").strip().lower()

        try: # If the player typed a number, check it's in range
            number = int(choice)
            if 1 <= number <= len(options):
                return options[number - 1]
            print("That number isn't an option. Please pick 1 to {}.".format(len(options)))
        except ValueError: # Not a number, so checks if they typed the option's name instead
            if choice in options:
                return choice
            print("Sorry, '{}' isn't a valid option. Please try again.".format(choice))
```

```python
def get_valid_guess(guessed_letters):
    """Ask the player for a letter and keep asking until it is a single,
    new letter from A to Z. Returns the guess in lower case."""
    while True:
        guess = input("Guess a letter: ").strip().lower()

        if len(guess) == 0:
            print("You didn't type anything - please enter a letter.")
        elif len(guess) > 1:
            print("Please enter only ONE letter at a time.")
        elif not guess.isalpha():
            print("That's not a letter - numbers and symbols aren't allowed.")
        elif guess in guessed_letters:
            print("You've already guessed '{}' - try a different letter.".format(guess))
        else:
            return guess
```

5. Statistics functions

```python
def update_stats(won, word, category, difficulty, guesses_used):
    """Update the global counters and add this game to the game history."""
    global games_played, wins, losses, current_streak, best_streak

    games_played += 1

    if won: # Checks if this is the player's best streak so far
        wins += 1
        current_streak += 1
        result = "Win"
        if current_streak > best_streak:
            best_streak = current_streak
    else:
        losses += 1
        current_streak = 0  # a loss resets the streak
        result = "Loss"

    # Stored as a record of this game as a dictionary inside the history list
    game_history.append({
        "game": games_played,
        "word": word,
        "category": category,
        "difficulty": difficulty,
        "result": result,
        "guesses": guesses_used
    })
```

```python
def show_stats():
    """Print the player's statistics so far.""" 
    if games_played > 0:
        win_rate = wins / games_played * 100
    else:
        win_rate = 0

    print("\n----- YOUR STATS -----")
    print("Games played:    ", games_played)
    print("Wins:            ", wins)
    print("Losses:          ", losses)
    print("Win rate:         %.1f%%" % win_rate)
    print("Correct guesses: ", correct_guesses)
    print("Wrong guesses:   ", wrong_guesses)
    print("Current streak:  ", current_streak)
    print("Best streak:     ", best_streak)
    print("----------------------")
```

```python
def show_final_summary():
    """Print the final stats and a record of every game played."""
    print("\n========== FINAL SUMMARY ==========")
    show_stats()
    print("\nGame history:")
    for game in game_history:
        print("  Game {}: {} ({}, {}) - {} in {} guesses".format(
            game["game"], game["word"].upper(), game["category"],
            game["difficulty"], game["result"], game["guesses"]))
    print("\nThanks for playing Hangman!")
```

6. Run the game

```python
def main():
    """Run the game: show instructions, play rounds until the player
    chooses to stop, then show the final summary."""
    show_instructions()

    while True:
        category = get_choice("Choose a category:", list(WORD_BANK.keys()))
        difficulty = get_choice("Choose a difficulty:", list(DIFFICULTY_LIVES.keys()))

        word = get_random_word(category)
        lives = DIFFICULTY_LIVES[difficulty]
        print("\nThe word has {} letters. Good luck!".format(len(word)))

        # play_round returns two values, so it's unpacked into two variables
        won, guesses_used = play_round(word, lives)
        update_stats(won, word, category, difficulty, guesses_used)
        show_stats()

        play_again = get_choice("Do you want to play again?", ["yes", "no"])
        if play_again == "no":
            break

    show_final_summary()


main()
```
